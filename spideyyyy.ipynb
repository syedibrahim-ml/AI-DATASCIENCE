```python
# Task 2 - Mobile Phone Data Usage Analyzer

# Days ki list banayein
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
data_usage = []

print("--- Mobile Phone Data Usage Analyzer ---")
print("Please enter the GB used for each day:\n")

# User se har din ka data input lena
for day in days:
    while True:
        try:
            usage = float(input(f"Enter GB used on {day}: "))
            if usage >= 0:
                data_usage.append(usage)
                break
            else:
                print("Usage negative nahi ho sakti. Dubara enter karein.")
        except ValueError:
            print("Invalid input! Sirf numbers enter karein.")

# 1. Total Weekly Usage ki calculation
total_weekly_usage = sum(data_usage)

# 2. Average Daily Usage ki calculation
average_daily_usage = total_weekly_usage / len(data_usage)

# 3. Highest Daily Usage nikalna
highest_daily_usage = max(data_usage)

# 4. Heavy-usage days count karna (GB > 4.0 ko heavy maana hai)
heavy_usage_days = sum(1 for usage in data_usage if usage > 4.0)

# 5. Weekly Category decide karna (Total usage ke mutabiq)
if total_weekly_usage < 15.0:
    weekly_category = "Light"
elif total_weekly_usage <= 25.0:
    weekly_category = "Moderate"
else:
    weekly_category = "Heavy"

# Results ko display karna
print("\n" + "="*35)
print("          ANALYSIS REPORT          ")
print("="*35)
print(f"Total Weekly Usage:      {total_weekly_usage:.1f} GB")
print(f"Average Daily Usage:     {average_daily_usage:.2f} GB")
print(f"Highest Daily Usage:     {highest_daily_usage:.1f} GB")
print(f"Number of Heavy Days:    {heavy_usage_days}")
print(f"Weekly Category:         {weekly_category}")
print("="*35)
```

```python
# Task: Feedback Analyzer

print("*************************************")
print("          FEEDBACK ANALYZER          ")
print("*************************************")

# User se feedback input lena
feedback = input("Enter the customer feedback text:\n")

# 1. Basic Stats: Characters, Words, aur Vowels count karna
char_count = len(feedback)

# split() use karke words ki list banana
words_list = feedback.split()
word_count = len(words_list)

# Vowels count karne ka logic
vowels = "aeiouAEIOU"
vowel_count = sum(1 for char in feedback if char in vowels)

# 2. Feedback Category nikalna (diye gaye table ke mutabiq)
# Check karna ke text mein 'good' ya 'bad' maujood hain ya nahi
has_good = "good" in feedback.lower()
has_bad = "bad" in feedback.lower()

if has_good and not has_bad:
    category = "Positive"
elif has_bad and not has_good:
    category = "Negative"
elif has_good and has_bad:
    category = "Mixed"
else:
    category = "Neutral"

# 3. Longest Word dhoondna (Loop ke zariye)
longest_word = ""
for word in words_list:
    # Punctuation (jaise '.', ',') ko clean karna taake sirf word ki length check ho
    cleaned_word = word.strip(".,!?;:")
    if len(cleaned_word) > len(longest_word):
        longest_word = cleaned_word

# Results ko display karna
print("\n" + "-"*35)
print(f"Characters: {char_count}")
print(f"Words:      {word_count}")
print(f"Vowels:     {vowel_count}")
print(f"Category:   {category}")
print(f"Longest Word: '{longest_word}'")
print("-"*35)
```

```python
# Task 1 - Supermarket Checkout System

print("--- Supermarket Checkout System ---\n")

# 1. Sabse pehle poochhein kitne different products hain
try:
    num_products = int(input("How many products? "))
except ValueError:
    print("Invalid input! Please enter a valid number.")
    num_products = 0

# Products ka data store karne ke liye list
purchased_items = []
grand_total = 0

# 2. Har product ke liye loop chalayein aur details lein
for i in range(1, num_products + 1):
    print(f"\nProduct {i}")
    name = input("Name: ")

    try:
        price = float(input("Price: "))
        quantity = int(input("Quantity: "))
    except ValueError:
        print("Invalid number input! Setting price and quantity to 0.")
        price = 0.0
        quantity = 0

    # Is item ka total cost
    item_total = price * quantity
    grand_total += item_total

    # Data ko list mein save karna (taki end mein display kar sakein)
    purchased_items.append({
        "name": name,
        "price": price,
        "quantity": quantity,
        "total": item_total
    })

# 3. Final Receipt/Bill display karna
print("\n" + "="*40)
print("             FINAL RECEIPT            ")
print("="*40)
for idx, item in enumerate(purchased_items, 1):
    print(f"{idx}. {item['name']} x {item['quantity']} | Price: {item['price']:.2f} | Total: {item['total']:.2f}")

print("-"*40)
print(f"GRAND TOTAL: Rs. {grand_total:.2f}")
print("="*40)
```

```python
# Task: Employee Attendance System

print("--- Employee Attendance System ---\n")

# Total 5 employees ke liye system
total_employees = 5
working_days = 20

# 2D list jisme data store hoga: [[Name, Present, Absent], ...]
attendance_data = []

# 1. 5 Employees ka data user se input lena
for i in range(1, total_employees + 1):
    print(f"\nEnter details for Employee {i}:")
    name = input("Employee Name: ")

    while True:
        try:
            present = int(input("Present days: "))
            absent = int(input("Absent days: "))

            # Validation: Check karna ke total days 20 se zyada na hon
            if (present + absent) <= working_days and present >= 0 and absent >= 0:
                break
            else:
                print(f"Error! Total days (Present + Absent) 20 se zyada nahi ho sakte. Dubara enter karein.")
        except ValueError:
            print("Invalid input! Sirf numbers enter karein.")

    # 2D list mein data append karna
    attendance_data.append([name, present, absent])

# 2. Calculations aur Report Display karna
print("\n" + "="*55)
print("                    ATTENDANCE REPORT               ")
print("="*55)
print(f"{'Name':<12} | {'Present':<7} | {'Absent':<6} | {'Percentage':<10} | {'Status':<10}")
print("-"*55)

for emp in attendance_data:
    emp_name = emp[0]
    emp_present = emp[1]
    emp_absent = emp[2]

    # Formula ke mutabiq percentage nikalna
    percentage = (emp_present / working_days) * 100

    # Table ke mutabiq Status decide karna
    if percentage >= 90:
        status = "Excellent"
    elif 75 <= percentage <= 89:
        status = "Good"
    else:
        status = "Warning"

    # Formatting ke sath print karna
    print(f"{emp_name:<12} | {emp_present:<7} | {emp_absent:<6} | {percentage:.1f}%      | {status:<10}")

print("="*55)
```
